# RAG System Demo
Query the Ocean ML RAG API.

```python
import requests


API = 'http://localhost:8000'
print('API:', API)
```

```python
def ask(question):
    r = requests.post(
        f"{API}/ask",
        json={"question": question, "top_k": 3}
    )

    return r.json()
```

```python
questions = [
    "Which datasets contain sea water temperature?",
    "Which source files contributed to sea water temperature?",
    "What variables exist in the ocean datasets?"
]
```

```python
for q in questions:
    print('\n===========================')
    print('QUESTION:', q)
    res = ask(q)
    print('\nANSWER:')
    print(res['answer'])
```
